## Sender Balance Analysis

We compare the sender's balance before and after the transaction to identify differences between legitimate and fraudulent transactions.

```python
financial_transaction_data.groupby("isFraud")[["oldbalanceOrg", "newbalanceOrig"]].mean()
```

```python
balance_change = financial_transaction_data["oldbalanceOrg"] - financial_transaction_data["newbalanceOrig"]

financial_transaction_data["balance_error"] = balance_change - financial_transaction_data["amount"]

financial_transaction_data["balance_error"].describe()
```

```python
financial_transaction_data.groupby("isFraud")["balance_error"].agg(["mean", "median", "min", "max"])
```

### Receiver Balance Analysis

```python
''financial_transaction_data.head()
```

```python
financial_transaction_data["dest_balance_error"]=(financial_transaction_data["newbalanceDest"]-financial_transaction_data["oldbalanceDest"]-financial_transaction_data["amount"])
financial_transaction_data["dest_balance_error"]
```

```python
financial_transaction_data.groupby("isFraud")["dest_balance_error"].agg(["mean", "median", "min", "max"])
```
